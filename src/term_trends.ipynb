```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_context("paper", font_scale=1.25)
sns.set_style("white")
degradomics = pd.read_csv("../data/degradomics_2025_10_02.csv")
peptidomics = pd.read_csv("../data/peptidomics_2025_10_02.csv")

degradomics["Year"] = pd.to_datetime(degradomics["Year"], format='%Y')
peptidomics["Year"] = pd.to_datetime(peptidomics["Year"], format='%Y')

fig, ax1 = plt.subplots(figsize=(3,2.5))

ax1.bar(peptidomics["Year"].dt.year, peptidomics["Count"], color="royalblue", alpha=0.6, label="Peptidomics")
ax1.set_xlabel("Year")
ax1.set_ylabel("Peptidomics")
ax1.tick_params(axis='y', labelcolor="royalblue")

ax2 = ax1.twinx()
ax2.bar(degradomics["Year"].dt.year, degradomics["Count"], color="teal", alpha=0.6, label="Degradomics")
ax2.set_ylabel("Degradomics")
ax2.tick_params(axis='y', labelcolor="teal")


plt.xticks(rotation=45)
ax1.grid(axis='y', linestyle="--", alpha=0.5)

sns.despine(top=True, right=False)
plt.savefig("../figures/term_trends.svg")
```
